```python
import sys
import os
import matplotlib.pyplot as plt  # Missing import!

current_dir = os.getcwd()
project_root = os.path.dirname(current_dir)
sys.path.insert(0, project_root)
from src.batch_processor import BatchProcessor

# Initialize batch processor
processor = BatchProcessor()

# Process entire dataset
results = processor.process_dataset("../MICC-F220/")
```

```python
# Evaluate performance
metrics = processor.evaluate_performance(results)

print("Performance Metrics:")
print(f"Precision: {metrics['precision']:.3f}")
print(f"Recall: {metrics['recall']:.3f}")
print(f"F1-Score: {metrics['f1_score']:.3f}")
print(f"Accuracy: {metrics['accuracy']:.3f}")
```

```python
# Generate comprehensive report
report_path = processor.generate_report(results, "../results/")
print(f"Report generated: {report_path}")
print("Check the results/ folder for:")
print("  - detection_report.json")
print("  - detection_summary.csv")
print("  - detection_report.txt")
print("  - statistics.png")
print("  - performance_metrics.png")
print("  - sample_detections/")
```

```python
# Analyze false positives and false negatives
false_positives = []
false_negatives = []

ground_truth = processor._infer_ground_truth_micc(results)

for result in results:
    filename = os.path.basename(result["image_path"])
    predicted = result["is_forgery"]
    actual = ground_truth.get(filename, False)

    if predicted and not actual:
        false_positives.append(result)
    elif not predicted and actual:
        false_negatives.append(result)

print(f"False Positives: {len(false_positives)}")
print(f"False Negatives: {len(false_negatives)}")

# Show examples of errors
if false_positives:
    print("\nFalse Positive Examples:")
    for fp in false_positives[:3]:
        print(f"  - {os.path.basename(fp['image_path'])}")

if false_negatives:
    print("\nFalse Negative Examples:")
    for fn in false_negatives[:3]:
        print(f"  - {os.path.basename(fn['image_path'])}")
```
